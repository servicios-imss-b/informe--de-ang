```python
clues_base_ang = set(base_an['clues_imb'].dropna().astype(str).str.strip().str.upper())
clues_base_arm = set(base_ar['clues_imb'].dropna().astype(str).str.strip().str.upper())

existen_en_ambas = pd.DataFrame({
    'clues_imb': sorted(clues_base_ang & clues_base_arm),
    'resultado': 'Existen en ambas'
})
solo_en_base_arm = pd.DataFrame({
    'clues_imb': sorted(clues_base_arm - clues_base_ang),
    'resultado': 'En base_arm'
})
solo_en_base_ang = pd.DataFrame({
    'clues_imb': sorted(clues_base_ang - clues_base_arm),
    'resultado': 'En base_ang'
})

comparacion_clues = pd.concat(
    [existen_en_ambas, solo_en_base_arm, solo_en_base_ang],
    ignore_index=True
)

print(f'Existen en ambas: {len(existen_en_ambas)}')
print(f'Solo en base_arm (base_ar): {len(solo_en_base_arm)}')
print(f'Solo en base_ang (base_an): {len(solo_en_base_ang)}')
print(f'Total de CLUES distintas: {len(comparacion_clues)}')
comparacion_clues
```
